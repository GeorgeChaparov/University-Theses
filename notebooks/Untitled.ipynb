```python
from pandas import read_csv

file_path = "./labels.csv"
df = read_csv(file_path)

# remove first row
df = df.iloc[1:].copy()

# subtract 750
df["start timestamp [ms]"] = df["start timestamp [ms]"] - 750
df["end timestamp [ms]"] = df["end timestamp [ms]"] - 750

df.iloc[0, 1] = 0
print(df)
output_path = "../data/Training Data/labels.csv"
df.to_csv(output_path, index=False)

output_path
```
